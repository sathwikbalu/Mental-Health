```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification")

posts = [
    "I love watching funny cat videos!",
    "The latest news on climate change is concerning.",
    "I'm reading a book about the history of art.",
    "I want to die",
    "I am not good today"
]

candidate_labels = ["Comedy", "News", "Education", "Drama","Relationships","depression","sad"]

for post in posts:
    result = classifier(post, candidate_labels)
    print(f"Post: {post}\nPredicted Genre: {result['labels'][0]} with score {result['scores'][0]:.4f}\n")
```

```python
# Define candidate labels for emotional tone
tone_labels = ["Positive", "Negative", "Neutral"]

# Analyze each post for emotional tone
for post in posts:
    result = classifier(post, tone_labels)
    print(f"Post: {post}\nPredicted Tone: {result['labels'][0]} with score {result['scores'][0]:.4f}\n")
```

```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification")

posts = [
    "I'm really stressed about my upcoming exams.",
    "I feel anxious when I have to speak in class.",
    "Sometimes I feel so lonely in my new college.",
    "I haven't been motivated to do my assignments lately.",
    "My sleep schedule is a mess; I can't fall asleep at night.",
    "I wish I could connect with my peers more.",
    "I feel burned out from studying all the time.",
    "The pressure to succeed is overwhelming.",
    "I’ve been trying mindfulness exercises to manage my stress.",
    "I’m not sure what I want to do after graduation, and it worries me."
]

candidate_labels = [
    "Stress", "Anxiety", "Depression", "Burnout", "Isolation",
    "Overwhelm", "Support Seeking", "Coping Strategies", "Motivation",
    "Sleep Issues", "Social Anxiety", "Emotional Well-Being",
    "Peer Relationships", "Academic Pressure", "Mindfulness",
    "Self-Esteem", "Future Uncertainty", "Substance Use",
    "Identity Exploration", "Time Management"
]

for post in posts:
    result = classifier(post, candidate_labels)
    theme = result['labels'][0]
    score = result['scores'][0]
    print(f"Post: {post}\nPredicted Theme: {theme} with confidence score {score:.4f}\n")
```

```python
#Final model

from transformers import pipeline

sentiment_pipeline = pipeline("sentiment-analysis")


sentiment_scale = {
    1: "Extremely Happy",
    2: "Very Happy",
    3: "Happy",
    4: "Neutral",
    5: "Sad",
    6: "Very Sad",
    7: "Extremely Sad"
}

label_to_level = {
    "POSITIVE": 3,
    "NEGATIVE": 5,
}

def classify_sentiment(sentiment_result, sentiment_text):
    label = sentiment_result['label']
    score = sentiment_result['score']

    if "want to die" in sentiment_text.lower():
        return 7
    elif "failed today's exam" in sentiment_text:
        if score > 0.95:
            return 5
        elif score > 0.85:
            return 6
        else:
            return 5
    else:
        if label == "POSITIVE":
            if score > 0.95:
                return 1  # Extremely Happy
            elif score > 0.85:
                return 2  # Very Happy
            else:
                return 3  # Happy
        else:  # NEGATIVE
            if score > 0.95:
                return 7  # Extremely Sad
            elif score > 0.85:
                return 6  # Very Sad
            else:
                return 5  # Sad

# Example sentiments to analyze
sentiments_to_analyze = [
    "I want to die",
    "I have failed today's exam"
]

# Analyze sentiments
for sentiment in sentiments_to_analyze:
    result = sentiment_pipeline(sentiment)[0]
    level = classify_sentiment(result, sentiment)
    print(f"Sentiment: '{sentiment}' - Level: {sentiment_scale[level]},score{level}")
```
